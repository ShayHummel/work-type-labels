# Zuzai work-type labels — embeddings on Colab (resumable)

**How to run:** Runtime → Change runtime type → GPU (A100 or L4 preferred; T4 skips the 8B model) → **Runtime → Run all**.

- All progress is saved to Google Drive (`My Drive/zuzai-hw/`). If the GPU is taken away, reconnect and **Run all** again: finished chunks and models are skipped.
- When it finishes, download the zips from `My Drive/zuzai-hw/exports/` and on the Mac run
  `uv run python src/import_colab.py ~/Downloads/emb_*.zip`
- Logs: `My Drive/zuzai-hw/logs/`.
